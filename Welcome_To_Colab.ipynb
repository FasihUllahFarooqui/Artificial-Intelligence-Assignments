```
flights = [
    {"id": "FL101", "from": "New York", "to": "London", "price": 450},
    {"id": "FL102", "from": "New York", "to": "London", "price": 520},
    {"id": "FL201", "from": "London", "to": "Tokyo", "price": 800},
    {"id": "FL301", "from": "Paris", "to": "New York", "price": 380},
]

departure_city = input("Enter departure city: ").strip()
destination_city = input("Enter destination city: ").strip()

matching_flights = [
    flight
    for flight in flights
    if flight["from"].lower() == departure_city.lower()
    and flight["to"].lower() == destination_city.lower()
]

if not matching_flights:
    print("No flights available")
else:
    print("\nAvailable Flights:")
    for flight in matching_flights:
        print(f"ID: {flight['id']} | Price: ${flight['price']}")

    selected_id = input("\nEnter flight ID: ").strip()
    print("Flight booked successfully")
```
